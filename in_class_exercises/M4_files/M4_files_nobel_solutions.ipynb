Some entries have **no `laureates` key at all**, and a `topMotivation` that says no prize was awarded. Real data has holes like this, and this one is easy to miss:
- `prize["laureates"]` raises a `KeyError` on these entries
- `prize.get("laureates", [])` returns an empty list instead

**Your turn:** build `awarded_by_year`, a dictionary mapping each year to the **number of prizes actually awarded** (prizes with at least one laureate). Every year should appear, even if few prizes were awarded. Then make `short_years`, a list of the years with fewer than 5.

````python
awarded_by_year = {}
for prize in prizes:
    year = prize["awardYear"]
    if year not in awarded_by_year:
        awarded_by_year[year] = 0
    if prize.get("laureates"):
        awarded_by_year[year] += 1

short_years = []
for year, n in awarded_by_year.items():
    if n < 5:
        short_years.append(year)

print(short_years)
awarded_by_year
````

````python
# Check
assert len(awarded_by_year) == 25, "every year should appear"
assert sum(awarded_by_year.values()) == 106
assert awarded_by_year["1916"] == 1
````

Watch out:

### 3. Write a function
Write `write_unawarded(input_filename, output_filename)`. It reads the Nobel JSON file and writes one line for each prize that was **not** awarded, like `1916: Chemistry`, to a text file. It returns the number of lines it wrote.

Use the pattern: load, transform, write. Sketch it on paper first.

````python
def write_unawarded(input_filename, output_filename):
    '''
    Write one line per unawarded Nobel Prize, like "1916: Chemistry".

    Inputs:
      input_filename: (string) name of the Nobel JSON file
      output_filename: (string) name for the output text file

    Returns: (int) the number of lines written
    '''
    # Load
    with open(input_filename) as f:
        prizes = json.load(f)["nobelPrizes"]

    # Transform
    lines = []
    for prize in prizes:
        if not prize.get("laureates"):
            lines.append(f"{prize['awardYear']}: {prize['category']['en']}")

    # Write
    with open(output_filename, "w") as f:
        for line in lines:
            print(line, file=f)

    return len(lines)
````

````python
# Check
n = write_unawarded("nobelPrizes.json", "unawarded.txt")
assert n == 19
with open("unawarded.txt") as f:
    first = f.readline()
assert first == "1914: Literature\n", repr(first)
````

### 4. Helper functions and comprehensions

Big functions are easier to write and test when they're built from small **helper functions** that each do one thing. You write and test the helpers first, then the bigger function just calls them (like the leaf functions on the dictionaries worksheet).

A **list comprehension** builds a list in one line. These two do the same thing:

```python
evens = []
for n in nums:
    if n % 2 == 0:
        evens.append(n * 10)

evens = [n * 10 for n in nums if n % 2 == 0]
```

**4a.** Write two helpers:
- `is_awarded(prize)`: `True` if the prize has at least one laureate, `False` otherwise
- `laureate_name(laureate)`: the laureate's English `knownName`, or the English `orgName` for an organization

````python
def is_awarded(prize):
    '''
    Was this prize awarded?

    Inputs:
      prize: (dict) one prize from the Nobel data

    Returns: (bool) True if it has at least one laureate
    '''
    return len(prize.get("laureates", [])) > 0


def laureate_name(laureate):
    '''
    The English name of a laureate (a person or an organization).

    Inputs:
      laureate: (dict) one laureate from a prize's "laureates" list

    Returns: (string) the knownName, or the orgName for an organization
    '''
    if "knownName" in laureate:
        return laureate["knownName"]["en"]
    return laureate["orgName"]["en"]
````

````python
# Check: test each helper on its own before using it
assert is_awarded(prizes[0]) is True
assert is_awarded(first_1916) is False
assert laureate_name({"knownName": {"en": "Marie Curie"}}) == "Marie Curie"
assert laureate_name({"orgName": {"en": "International Committee of the Red Cross"}}) == "International Committee of the Red Cross"
````

**4b.** Now use the helpers:
- `prize_names(prize)`: a list of the names of everyone who won this prize. Use a **list comprehension** and `laureate_name`. An unawarded prize should give `[]`.
- `winners_in(prizes, category)`: a list of the names of everyone who won in `category`, in order. Call `prize_names`; don't repeat its work.

````python
def prize_names(prize):
    '''
    Names of everyone who won this prize.

    Inputs:
      prize: (dict) one prize from the Nobel data

    Returns: (list of strings) names; [] if the prize wasn't awarded
    '''
    return [laureate_name(laureate) for laureate in prize.get("laureates", [])]


def winners_in(prizes, category):
    '''
    Names of everyone who won a prize in one category.

    Inputs:
      prizes: (list of dicts) the Nobel prizes
      category: (string) e.g. "Peace"

    Returns: (list of strings) names, in order
    '''
    names = []
    for prize in prizes:
        if prize["category"]["en"] == category:
            names.extend(prize_names(prize))
    return names
````

````python
# Check
assert prize_names(first_1916) == []
assert prize_names(prizes[13]) == ["Henri Becquerel", "Pierre Curie", "Marie Curie"]   # 1903 Physics
peace = winners_in(prizes, "Peace")
assert peace[:2] == ["Henry Dunant", "Frédéric Passy"]
assert "International Committee of the Red Cross" in peace
print(len(peace), "Peace laureates, 1901-1925")
````
