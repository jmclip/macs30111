## 4. Adding and removing elements

````python
lst = [1, 2, 3]
print("start:", lst, "id:", id(lst))

# append() adds ONE element to the end, IN PLACE (returns None)
result = lst.append(4)
print("after append:", lst, "| append returned:", result, "| id:", id(lst))

# extend() adds every element of another list, IN PLACE
lst.extend([5, 6])
print("after extend:", lst, "| id:", id(lst))

# insert(position, value) adds at a specific position, IN PLACE
lst.insert(0, 100)
print("after insert:", lst, "| id:", id(lst))

# + creates a NEW list (notice the id changes)
new_lst = lst + [7, 8]
print("with +:", new_lst, "| id:", id(new_lst))
print("original unchanged:", lst)
````

## 5. List comprehensions

```
new = [<expr> for <var> in <list>]                             # transform
new = [<expr> for <var> in <list> if <condition>]              # filter
new = [<expr> if <condition> else <expr> for <var> in <list>]  # transform two ways
```

````python
lst = [1, 2, 3, 4]

# Transform -- the long way
doubled = []
for x in lst:
    doubled.append(x * 2)
print(doubled)

# Transform -- list comprehension
doubled2 = [x * 2 for x in lst]
print(doubled2)
````

````python
lst = [1, 2, 3, 4]

# Filter -- the long way
evens = []
for x in lst:
    if x % 2 == 0:
        evens.append(x)
print(evens)

# Filter -- list comprehension (if goes at the END)
evens2 = [x for x in lst if x % 2 == 0]
print(evens2)

# Transform two ways -- if/else goes at the FRONT
labels = ["even" if x % 2 == 0 else "odd" for x in lst]
print(labels)
````

````python
# SOLUTION
# Use nums = list(range(0, 61, 3)).
# Create a new list that SQUARES even numbers and makes odd numbers NEGATIVE.
# Partners: one of you does it the "long" way, the other with a list comprehension.
# Check your answers match with ==.

nums = list(range(0, 61, 3))

# Long way
result_long = []
for x in nums:
    if x % 2 == 0:
        result_long.append(x ** 2)
    else:
        result_long.append(-x)
print(result_long)

# List comprehension (if/else goes at the FRONT because we transform every element)
result_comp = [x ** 2 if x % 2 == 0 else -x for x in nums]
print(result_comp)

print(result_long == result_comp)
````

### 6. Working with lists: slicing and other operations

````python
lang = ['C', 'C++', 'Python', 'Java']

print(lang[1:3])    # positions 1 and 2 (stop is excluded)
print(lang[:2])     # omit start -> from the beginning
print(lang[2:])     # omit stop  -> to the end
print(lang[::2])    # step of 2

# A slice is a COPY: changing it doesn't change the original
part = lang[1:3]
part[0] = "Rust"
print("slice:", part, "| original:", lang)

# [:] copies the whole list
copy = lang[:]
print(copy is lang)
````

````python
nums = [4, 1, 7, 3, 7]

print(min(nums), max(nums), sum(nums))
print(nums.count(7))      # how many 7s
print(3 in nums)          # membership test

print(sorted(nums))       # returns a NEW sorted list
print(nums)               # original unchanged

nums.sort()               # sorts IN PLACE, returns None
print(nums)

nums.reverse()            # reverses IN PLACE
print(nums)
````

## 8. Tuples 

Work through these examples on your own. Run each cell, read the comments, and change things to see what happens.

**Creating tuples and indexing.** Tuples use parentheses and support indexing, slicing, `len`, and `in`, just like lists.

````python
tpl = (1, 2, 3, 4)
print(tpl[0], tpl[-1], tpl[1:3], len(tpl), 3 in tpl)

# A one-element tuple needs a trailing comma!
not_a_tuple = (5)
a_tuple = (5,)
print(type(not_a_tuple), type(a_tuple))

# The parentheses are actually optional -- the commas make the tuple
also_tuple = 1, 2, 3
print(type(also_tuple), also_tuple)
````

**Tuples are immutable.** Once created, you can't change, add, or remove values.

````python
tpl = (1, 2, 3, 4)
try:
    tpl[0] = 99
except TypeError as e:
    print("TypeError:", e)

print(hasattr(tpl, "append"))   # no append, extend, pop, remove...

# To "change" a tuple you build a new one
tpl2 = tpl + (5,)
print(tpl, tpl2)
````

**Unpacking.** Assign each value in a tuple to its own variable in one line.

````python
point = (3, 7)
x, y = point
print("x =", x, "y =", y)

# A handy trick: swapping two variables
a, b = 1, 2
a, b = b, a
print(a, b)
````

**Unpacking in a for loop.** Very common with lists of tuples (and you've already seen it with `enumerate`!).

````python
employees = [("Sam", "CEO"), ("Alex", "CTO"), ("Pat", "VP")]
for name, position in employees:
    print(name, "is the", position)

# enumerate() hands you (index, value) tuples -- that's why `for i, x in ...` works
print(list(enumerate(["a", "b", "c"])))
````

**Returning multiple values from a function** actually returns a tuple.

````python
def min_max(nums):
    return min(nums), max(nums)

result = min_max([4, 1, 9, 3])
print(result, type(result))

lo, hi = min_max([4, 1, 9, 3])
print("lo:", lo, "hi:", hi)
````

**Converting between lists and tuples.**

````python
lst = [1, 2, 3]
tpl = tuple(lst)
back = list(tpl)
print(tpl, back)
````

**Why use a tuple instead of a list?**
- The data is a fixed record that shouldn't change (a coordinate, a (name, position) pair, an RGB color).
- Immutability protects you from accidentally modifying it.
- Tuples can be used as dictionary keys; lists can't (more on dictionaries later!).

````python
locations = {(41.79, -87.60): "UChicago"}
print(locations[(41.79, -87.60)])

try:
    bad = {[41.79, -87.60]: "UChicago"}
except TypeError as e:
    print("TypeError:", e)
````

### Application How to fix this:

s=[[“Baby love, I   think I've been a little too.  Kind”], [“Didn't notice you walkin' all over my   peace of mind”], [“In the   shoes I gave you as a    present”]]



````python
# SOLUTION
# Clean up s: remove the stray periods, collapse the extra spaces,
# and put one period at the end of each line.
s=[["Baby love, I   think I've been a little too.  Kind"], 
   ["Didn't notice you walkin' all over my   peace of mind"], 
    ["In the   shoes I gave you as a    present"]]

final = []
for v in s:
    for line in v:
        no_periods = line.replace(".", "")          # remove every period
        clean = " ".join(no_periods.split())        # collapse extra spaces
        clean = clean[0].upper() + clean[1:]        # capitalize first letter only
        # Note: .capitalize() would also lowercase everything else, turning "I" into "i"!
        # Optional: .replace(" Kind", " kind") to drop the leftover capital from the removed period
        final.append(clean + ".")                   # re-add one period at the end

for line in final:
    print(line)
````

````python
# Reassigning a simple variable doesn't affect the other
a = 42
b = a
a = 37
print(a, b)

# s = r: both names point to the SAME list
r = [1, 2, 3, 4]
s = r
s[1] = 0
print(r, s, r is s)

# s = r[:]: s is a separate copy
r = [1, 2, 3, 4]
s = r[:]
s[1] = 0
print(r, s, r is s)
````

````python
m = [[0] * 5] * 5                  # 5 references to the SAME row
m1 = [[0] * 5 for i in range(5)]   # 5 separate rows

m[2][3] = 1
m1[2][3] = 1
print("m: ", m)
print("m1:", m1)
````

````python
# SOLUTION
# Challenge: can you use m1[:] to make a fully independent copy of m1?
# Make m2 = m1[:], change m2[0][0], and check m1. Explain what you see.

m1 = [[0] * 5 for i in range(5)]
m2 = m1[:]
m2[0][0] = 99
print(m1[0][0])     # 99 -- m1 changed too!
print(m2 is m1)     # False: the OUTER list is new...
print(m2[0] is m1[0])  # True: ...but the rows inside are the SAME lists

# So m1[:] is a shallow copy. To copy the rows too, copy each one:
m1 = [[0] * 5 for i in range(5)]
m3 = [row[:] for row in m1]
m3[0][0] = 99
print(m1[0][0], m3[0][0])   # 0 99

# (Python's copy module also has copy.deepcopy(m1) for this.)
````
