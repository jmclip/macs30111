## Slide examples: broadcasting (slides 47–54)
**The rule (slide 49):** (1) line the shapes up on the right, (2) a missing dimension on the left counts as 1, (3) each column must be equal, or one of them must be 1. Otherwise: `ValueError`.

```python
# Slide 47: a scalar stretches to every element
x = np.array([1, 2, 3])
x * 4
```

```python
# Slide 48: (4, 3) + (3,) -> the row is reused for every row
a = np.array([[0, 0, 0], [10, 10, 10], [20, 20, 20], [30, 30, 30]])
b = np.array([1, 2, 3])
a + b
```

```python
# Slide 49: check shapes with the rule, without building any arrays
print(np.broadcast_shapes((4, 3), (3,)))           # (3,) counts as (1, 3)
print(np.broadcast_shapes((256, 256, 3), (3,)))    # an image times one value per color
print(np.broadcast_shapes((8, 1, 6, 1), (7, 1, 5)))
```

```python
# Slide 50: (4, 3) + (4, 1) -> the column is reused for every column
c = np.array([1, 2, 3, 4]).reshape(4, 1)
print(c.shape)
a + c
```

```python
# Slide 51: (4, 1) + (3,) -> both stretch, result (4, 3)
col = np.array([0, 10, 20, 30]).reshape(4, 1)
row = np.array([1, 2, 3])
print(col + row)
# the same idea makes a multiplication table
np.arange(1, 4).reshape(3, 1) * np.arange(1, 4)
```

```python
# Slide 52: (4, 3) + (4,) fails: (4,) counts as (1, 4), and 3 vs 4 don't match
try:
    a + np.array([1, 2, 3, 4])
except ValueError as err:
    print('ValueError:', err)

# Fix: make it a column, shape (4, 1)
a + np.array([1, 2, 3, 4]).reshape(4, 1)
```

```python
# Slide 53: matching shapes != correct. Row means have shape (2,), so they line up with COLUMNS.
d = np.array([[1., 10.],
              [3., 30.]])
print(d.mean(axis=1))               # [ 5.5 16.5]: one mean per ROW
print(d - d.mean(axis=1))           # 10 - 16.5: row 1's mean, subtracted in row 0. No error!
print(d - d.mean(axis=1, keepdims=True))   # (2, 1): each row minus its own mean
```

## Broadcasting and standardizing example (4.2.7, 4.2.8)
The patients from slide 59: one row per patient, columns are height (inches) and weight (lbs).

**Predict the shape (or the error) before you run each line.** Broadcasting rule: compare shapes right to left; each pair must be equal, or one of them must be 1.

```python
patients = np.array([[72.0, 190.0],
                     [60.0, 124.0],
                     [65.0, 170.0],
                     [75.0, 225.0],
                     [50.0,  85.0]])
patients.shape
```

### Part A: broadcasting warm-up
1. `patients + 10` — what shape comes back?
2. Convert to metric in one line: multiply by `np.array([2.54, 0.4536])` (cm per inch, kg per lb). Shapes `(5, 2)` and `(2,)`: does it work?
3. `patients - patients.mean(axis=1)` — does it run? If not, why not, and how would you fix it? (And does subtracting a row's mean even make sense here?)

```python
# 1: a scalar broadcasts to every element -> shape (5, 2)
(patients + 10).shape
```

```python
# 2: (5, 2) * (2,) -> compare right to left: 2 == 2, then (2,) is treated as (1, 2) -> (5, 2)
#    Each column gets its own conversion factor
metric = patients * np.array([2.54, 0.4536])
metric.round(1)
```

```python
# 3: patients.mean(axis=1) has shape (5,). (5, 2) vs (5,): rightmost dims are 2 vs 5 -> ValueError
try:
    patients - patients.mean(axis=1)
except ValueError as e:
    print('ValueError:', e)
```

```python
# Fix: make the row means a column, shape (5, 1); (5, 2) vs (5, 1) broadcasts
row_means = patients.mean(axis=1, keepdims=True)   # or .reshape(-1, 1) or [:, np.newaxis]
print(row_means.shape)
patients - row_means
# It runs, but it's meaningless: averaging inches with pounds. That's why we standardize by COLUMN (axis=0).
```

### Part B: standardize by hand
4. Compute `mu = patients.mean(axis=0)` and `sigma = patients.std(axis=0)`. What are their shapes?
5. `z = (patients - mu) / sigma`. Check that each column now has mean 0 and standard deviation 1.
6. Standardize the **metric** version too. Are the z-scores the same? (`np.allclose`) Why?
7. Use a boolean mask: which patients are above average on **both** height and weight?

```python
# 4: one mean and one std per column -> shape (2,)
mu = patients.mean(axis=0)
sigma = patients.std(axis=0)
print(mu, sigma.round(3), mu.shape, sigma.shape)   # [64.4 158.8] [8.913 49.272] (2,) (2,)
```

```python
# 5: (5, 2) - (2,) and (5, 2) / (2,) both broadcast across rows
z = (patients - mu) / sigma
print(z.round(2))
print('means:', z.mean(axis=0).round(10))   # [0. 0.] (may print -0.)
print('stds: ', z.std(axis=0))             # [1. 1.]
```

```python
# 6: unit changes are multiplying by a constant; standardizing divides it right back out
z_metric = (metric - metric.mean(axis=0)) / metric.std(axis=0)
np.allclose(z, z_metric)   # True
```

```python
# 7: combine two boolean masks with & (parentheses required)
both_above = (z[:, 0] > 0) & (z[:, 1] > 0)
print(both_above)          # [ True False  True  True False]
patients[both_above]       # patients 0, 2, 3
```

### Part 4: writing tests, from incomplete to better
A test is useful only if it **fails on buggy code**. To check that, we need some buggy code. Here are the functions we'll test:
- `standardize_features`: the original (fails on constant columns, Part 1)
- `standardize_axis1`: the classmate's bug (Part 2)
- `standardize_forgot_sigma`: forgets to divide (quiz Q6, answer C)
- `standardize_features_v2`: a fixed version

```python
def standardize_forgot_sigma(data):
    '''Bug: subtracts the mean but forgets to divide by sigma.'''
    return data - data.mean(axis=0)


def standardize_features_v2(data):
    '''
    Standardize features to have mean 0.0 and standard deviation 1.0.
    Columns with zero standard deviation come back as all zeros.
    '''
    mu_vec = data.mean(axis=0)
    sigma_vec = data.std(axis=0)
    sigma_vec = np.where(sigma_vec == 0, 1, sigma_vec)   # don't divide by 0
    return (data - mu_vec) / sigma_vec


def run_test(test, f):
    '''Run one test on one function: PASS, FAIL (an assert was False) or ERROR (crashed).'''
    with warnings.catch_warnings():
        warnings.simplefilter('ignore')
        try:
            test(f)
            print(f'PASS   {test.__name__}({f.__name__})')
        except AssertionError as err:
            print(f'FAIL   {test.__name__}({f.__name__}): {err}')
        except Exception as err:
            print(f'ERROR  {test.__name__}({f.__name__}): {type(err).__name__}')
```

#### Test 1: incomplete
It only checks that the function runs and keeps the shape. **Predict:** which of the four functions pass?

```python
def test_1_runs(f):
    '''Incomplete: checks only that it runs on the survey and keeps the shape.'''
    z = f(survey)
    assert z.shape == survey.shape, 'shape changed'


for f in [standardize_features, standardize_axis1, standardize_forgot_sigma, standardize_features_v2]:
    run_test(test_1_runs, f)
```

`standardize_forgot_sigma` **passes** even though it's wrong. The test never looks at the numbers, so it can't catch a wrong number.

#### Test 2: better
It checks the property we actually want: every question has mean 0 and std 1. It uses `np.allclose`, not `==`, because of rounding errors.

```python
z = standardize_features(survey)
print(z.mean(axis=0))          # tiny rounding errors, not exactly 0
print(z.mean(axis=0) == 0)     # so == fails
print(np.allclose(z.mean(axis=0), 0))
```

```python
def test_2_properties(f):
    '''Better: on the 200 x 5 survey, every question has mean 0 and std 1.'''
    z = f(survey)
    assert z.shape == survey.shape, 'shape changed'
    assert np.allclose(z.mean(axis=0), 0), 'column means are not 0'
    assert np.allclose(z.std(axis=0), 1), 'column stds are not 1'


for f in [standardize_features, standardize_axis1, standardize_forgot_sigma, standardize_features_v2]:
    run_test(test_2_properties, f)
```

Better: it catches `forgot_sigma` (FAIL) and the classmate's bug (ERROR). But the original `standardize_features` still **passes**, and we know from Part 1 that it breaks. The survey never has a question that everyone answered the same, so the test never tries the case that breaks. A bigger dataset doesn't make a test better: the right edge cases do.

#### Test 3: complete, with edge cases
Each case is a small survey chosen to **trigger a problem** from the worksheet, plus an assert that says **what should happen**. Case 3 is a yes/no question, stored as a True/False array.

```python
def test_3_edge_cases(f):
    '''Complete: a hand-worked square survey, a question everyone answered the same, and a yes/no question.'''
    # Case 1 (Part 2, row D): a square survey, with an answer worked out by hand
    d = np.array([[1., 4.],
                  [3., 8.]])
    expected = np.array([[-1., -1.],
                         [ 1.,  1.]])
    assert np.allclose(f(d), expected), 'case 1: wrong answer on a square survey'

    # Case 2 (Part 1): a question everyone answered the same should come back as zeros, not nan
    d = np.array([[1., 5.], [2., 5.], [3., 5.]])
    z = f(d)
    assert not np.isnan(z).any(), 'case 2: constant column gave nan'
    assert np.allclose(z[:, 1], 0), 'case 2: constant column is not all zeros'

    # Case 3: a yes/no question stored as True/False. True counts as 1, False as 0.
    d = np.array([[True, False],
                  [False, False],
                  [True, True],
                  [True, False]])
    z = f(d)
    assert np.allclose(z.mean(axis=0), 0), 'case 3: column means are not 0'
    assert np.allclose(z.std(axis=0), 1), 'case 3: column stds are not 1'
    # every "yes" is above its question's average, every "no" below: compare to a True/False array
    assert np.array_equal(z > 0, d), 'case 3: yes answers should be above average'


for f in [standardize_features, standardize_axis1, standardize_forgot_sigma, standardize_features_v2]:
    run_test(test_3_edge_cases, f)
```

### Back to the survey: why this matters
Now that the helper is tested, we can trust what it tells us about the survey. Standardized answers let us compare across questions: a z of +2 means "unusually high **for that question**", whether the question's typical answer is 3 or 8.

```python
z = standardize_features_v2(survey)

# Respondent 0: raw answers vs. standardized answers
print('raw:         ', survey[0])
print('standardized:', z[0].round(2))

# Which respondent gives the most unusual answers overall? (largest average distance from typical)
unusual = np.abs(z).mean(axis=1)
r = unusual.argmax()          # argmax: the index of the largest value
print('most unusual respondent:', r)
print('raw:         ', survey[r])
print('standardized:', z[r].round(2))
```

Without standardizing, an answer of 8 looks "high" on every question. After standardizing, an 8 on Q4 (typical answer 8) is ordinary, and an 8 on Q5 (typical answer 3) is very unusual. If the helper had a bug, every conclusion here would be wrong, often with no error to warn us.

### Part 5: takeaways
- **No error doesn't mean correct.** A warning isn't an error, `nan` can appear with no warning at all, and broadcasting can match shapes by accident.
- An **incomplete test** checks that code runs. A **better test** checks the property you want. A **complete test** also feeds in the edge cases that break things: everyone giving the same answer, a square survey, yes/no questions, skipped answers.
- A test is only useful if it **fails on buggy code**. Try your tests on a broken version, too.
